```python
import pandas as pd
```

#### Formatting `data-stopfake.csv`

```python
# load and look through data

df = pd.read_csv('./data-collection/data-stopfake.csv')
df
```

```python
# drop useless column

df = df.drop(columns=['page'])
```

```python
# drop whitespaces in the beginning of the text

df['text'] = df['text'].str.strip()
```

```python
# drop rows that dont start with specified prefixes

prefixes = [
    'Відеофейк: ', 
    'Фейк: ', 
    'Маніпуляція: ', 
    'Фотофейк: '
]
df = df[df['text'].apply(lambda x: any(x.startswith(prefix) for prefix in prefixes))]
```

```python
# drop those prefixes from the text

for prefix in prefixes:  
    df.loc[:, 'text'] = df['text'].str.replace(prefix, '', regex=False)
```

```python
# add the 'label' column

df['label'] = False
```

```python
df
```

```python
# save formatted dataset

df.to_csv('./data-collection/data-stopfake-formatted.csv', index=False)
```

#### Formatting `data-voxukraine.csv`

```python
# load and look through data

df = pd.read_csv('./data-collection/data-voxukraine.csv')
df
```

```python
# drop whitespaces in the beginning of the text

df['text'] = df['text'].str.strip()
```

```python
# drop useless ending from text

df.loc[:, 'text'] = df['text'].str.replace(r'\s*. Випуск №\d+$', '', regex=True)
```

```python
# drop rows that dont start with specified prefixes

prefixes = [
    'ФЕЙК: ', 
    'МАНІПУЛЯЦІЯ: ', 
    'БЕЗПІДСТАВНО: ', 
    'НЕПРАВДА: ', 
    'КОНСПІРОЛОГІЯ: ', 
    'ВІДЕОФЕЙК: ', 
    'САТИРА: ', 
    'ФОТОФЕЙК: ', 
    'Фейки щодо системи охорони здоров’я: '
]
df = df[df['text'].apply(lambda x: any(x.startswith(prefix) for prefix in prefixes))]
```

```python
# drop those prefixes from the text

for prefix in prefixes:  
    df.loc[:, 'text'] = df['text'].str.replace(prefix, '', regex=False)
```

```python
# add the 'label' column

df['label'] = False
```

```python
df
```

```python
# save formatted dataset

df.to_csv('./data-collection/data-voxukraine-formatted.csv', index=False)
```

#### Formatting `data-kaggle.csv`

```python
df = pd.read_csv('./data-collection/data-kaggle.csv')
```

```python
df
```

```python
# drop texts from the source that has already been used

df = df[df['Link'] != 'https://voxukraine.org/category/voks-informue/']
```

```python
# drop useless columns

df = df.drop(columns=['Unnamed: 0', 'Link'])
```

```python
# rename columns

df = df.rename(columns={'Text': 'text', 'Label': 'label'})
```

```python
# remove colons from the beginning of some texts

df['text'] = df['text'].str.lstrip(': ')
```

```python
df
```

```python
# save formatted dataset

df.to_csv('./data-collection/data-kaggle-formatted.csv', index=False)
```

#### Merging 3 datasets (`data-stopfake.csv`, `data-voxukraine.csv`, `data-kaggle.csv`)

```python
# load all datasets

df1 = pd.read_csv('./data-collection/data-stopfake-formatted.csv')
df2 = pd.read_csv('./data-collection/data-voxukraine-formatted.csv')
df3 = pd.read_csv('./data-collection/data-kaggle-formatted.csv')
```

```python
# merge datasets

df = pd.concat([df1, df2, df3])
df = df.reset_index(drop=True)
```

```python
df
```

```python
# check the true/false news ratio

f'True: {len(df[df["label"] == True])}, False: {len(df[df["label"] == False])}'
```

```python
# save the final dataset

df.to_csv('./data-collection/data.csv', index=False)
```
